Support Vector Machine (SVM)

 is one of the most powerful and widely used supervised machine learning algorithms. While it can be used for regression (SVR), it is primarily famous for classification tasks.
 The Hyperplane: A line (in 2D), a plane (in 3D), or a hyperplane (in higher dimensions) that divides your classes.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import svm
x = np.array([1, 2.5, 4, 1.2, 7.5, 1.8, 8.5, 6, 9.2, 3.1, 5, 8.2, 6.5, 7.8, 8.5])
y = np.array([1.5, 3.5, 7, 2.1, 9, 1.1, 10.5, 8.5, 9.8, 3.5, 2.5, 9.1, 7.8, 7.2, 3])
training_X = np.stack((x, y)).T
training_y = [0, 0, 1, 0, 1, 0, 1, 1, 1, 0, 0, 1, 1, 1, 0]
clf = svm.SVC(kernel='linear', C=1.0)
clf.fit(training_X, training_y)
w = clf.coef_[0]
a = -w[0] / w[1]
XX = np.linspace(0, 13)
yy = a * XX - clf.intercept_[0] / w[1]
plt.figure(figsize=(8, 6))
plt.plot(XX, yy, 'r--', linewidth=2, label='Decision Boundary')
plt.scatter(training_X[:, 0], training_X[:, 1], c=training_y, cmap='viridis', s=70, edgecolor='k')
plt.xlabel('X Feature')
plt.ylabel('Y Feature')
plt.title('SVM Classification with Custom Points and Colors')
plt.legend()
plt.show()
```
