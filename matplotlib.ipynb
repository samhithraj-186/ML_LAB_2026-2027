Import Matplotlib

```python
import matplotlib.pyplot as plt
```

Simple Line Plot

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)

plt.show()
```

Add Title and Labels

```python
plt.plot(x, y)

plt.title("My First Graph")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

Line Plot

```python
import pandas as pd
data = {
    "Name": ["Ali", "Rahul", "John", "Ravi", "Sai", "Jack"],
    "Age": [20, 21, 22, 23, 20, 21],
    "Marks": [85, 90, 78, 92, 89, 65]
}
df = pd.DataFrame(data)
plt.plot(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

Change Line Style

```python
plt.plot(df["Name"], df["Marks"], linestyle="--")

plt.show()
```

Dotted line

```python
plt.plot(df["Name"], df["Marks"], linestyle=":")

plt.show()
```

Different markers

```python
plt.plot(df["Name"], df["Marks"], marker="o")

plt.show()
```

Line Width

```python
plt.plot(df["Name"], df["Marks"], linewidth=3)

plt.show()
```

Figure Size

```python
plt.figure(figsize=(10, 5))

plt.plot(df["Name"], df["Marks"])

plt.title("Student Marks")

plt.show()
```

X-axis and Y-axis Limits

```python
plt.plot(df["Age"], df["Marks"])

plt.xlim(19, 24)
plt.ylim(50, 100)

plt.show()
```

Add Text to Graph

```python
plt.plot(df["Name"], df["Marks"])

plt.text(0, 85, "85")

plt.show()
```

Save Your Graph

```python
plt.plot(df["Name"], df["Marks"])

plt.savefig("student_marks.png")

plt.show()
```

Bar Chart

```python
plt.bar(df["Name"], df["Marks"])

plt.title("Student Marks")
plt.xlabel("Name")
plt.ylabel("Marks")

plt.show()
```

Scatter Plot

```python
plt.scatter(df["Age"], df["Marks"])

plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")

plt.show()
```

Histogram

```python
plt.hist(df["Marks"])

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

Pie Chart

```python
cities = ["Hyderabad", "Delhi", "Mumbai"]
students = [40, 30, 30]

plt.pie(students, labels=cities)

plt.title("Students by City")

plt.show()
```

Grid

```python
plt.plot(df["Name"], df["Marks"])

plt.grid()

plt.show()
```

Subplots

```python
plt.subplot(2, 2, 1)

plt.plot(df["Name"], df["Marks"])
plt.title("Line")

plt.subplot(2, 2, 2)

plt.bar(df["Name"], df["Marks"])
plt.title("Bar")

plt.subplot(2, 2, 3)

plt.scatter(df["Age"], df["Marks"])
plt.title("Scatter")

plt.subplot(2, 2, 4)

plt.hist(df["Marks"])
plt.title("Histogram")

plt.tight_layout()

plt.show()
```

Graphs

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("sampledata.csv")

# Create 2 rows × 3 columns
plt.figure(figsize=(15, 8))

# 1. Bar Chart
plt.subplot(2, 3, 1)
plt.bar(df["Name"], df["Marks"])
plt.title("Bar Chart")
plt.xlabel("Name")
plt.ylabel("Marks")

# 2. Line Chart
plt.subplot(2, 3, 2)
plt.plot(df["Name"], df["Marks"], marker="o")
plt.title("Line Chart")
plt.xlabel("Name")
plt.ylabel("Marks")

# 3. Scatter Plot
plt.subplot(2, 3, 3)
plt.scatter(df["Age"], df["Marks"])
plt.title("Scatter Plot")
plt.xlabel("Age")
plt.ylabel("Marks")

# 4. Histogram
plt.subplot(2, 3, 4)
plt.hist(df["Marks"], bins=5)
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")

# 5. Pie Chart
plt.subplot(2, 3, 5)
plt.pie(df["Marks"], labels=df["Name"], autopct="%1.1f%%")
plt.title("Pie Chart")

# Automatically adjust spacing
plt.tight_layout()

# Display all graphs
plt.show()
```
