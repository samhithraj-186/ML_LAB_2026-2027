Groupby

```python
import pandas as pd

data = {
    "Name": ["Ali", "Rahul", "John", "Ravi", "Sai", "Jack"],
    "Age": [20, 21, 22, 23, 20, 21],
    "Marks": [85, 90, 78, 92, 89, 65]
}

df = pd.DataFrame(data)

result = df.groupby("Age")["Marks"].mean()

print(result)
```
